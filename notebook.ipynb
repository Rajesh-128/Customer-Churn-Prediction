```python
from sklearn.linear_model import LogisticRegression

# Create a valid numeric target
if df["Churn"].dtype == "object":
    y = df["Churn"].str.strip().str.lower().map({"yes": 1, "no": 0})
else:
    y = pd.to_numeric(df["Churn"], errors="coerce").astype("Int64")

# Remove rows with invalid target values
valid_rows = y.notna()
df_clean = df.loc[valid_rows].copy()
y = y.loc[valid_rows].astype(int)

# Build features without target or duplicate target columns
X = df_clean.drop(
    columns=["Churn", "Churn Flag", "customerID"],
    errors="ignore"
)

X["TotalCharges"] = pd.to_numeric(
    X["TotalCharges"],
    errors="coerce"
).fillna(0)

X = pd.get_dummies(X, drop_first=True).astype(float)

# Split and scale
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train the model
log_model = LogisticRegression(max_iter=1000)
log_model.fit(X_train_scaled, y_train)

# Predictions
y_pred = log_model.predict(X_test_scaled)

print("Missing target values:", y.isna().sum())
print("Training accuracy:", log_model.score(X_train_scaled, y_train))
print("Test accuracy:", log_model.score(X_test_scaled, y_test))
```

```python
# Generate predicted probabilities for ROC-AUC
y_pred_proba = log_model.predict_proba(X_test_scaled)[:, 1]

# Ensure predictions are available
y_pred = log_model.predict(X_test_scaled)

print("Predictions generated successfully.")
print("y_pred shape:", y_pred.shape)
print("y_pred_proba shape:", y_pred_proba.shape)
```

```python
# Make predictions
y_pred = log_model.predict(X_test_scaled)

# Get probability of Churn = 1
y_pred_proba = log_model.predict_proba(X_test_scaled)[:, 1]

print("Predictions created successfully!")
```

```python
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score

accuracy = accuracy_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_pred_proba)

print("Accuracy:", accuracy)
print("ROC-AUC:", roc_auc)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
